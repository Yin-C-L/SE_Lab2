PyTorch 是一个 Python 库，主要提供两个高级功能：
- GPU 加速的张量（Tensor）计算
- 构建在反向自动求导系统上的深度神经网络

## 1. 定义数据

```python
import torch

# 标量
a = torch.tensor(666)
print(a)

# 向量
b = torch.tensor([1, 2, 3, 4, 5, 6])
print(b)

# 矩阵（3 行 3 列，全 1）
c = torch.ones(3, 3)
print(c)

# 三维张量（2 × 3 × 4，全 1）
d = torch.ones(2, 3, 4)
print(d)
```

Tensor 支持多种数据类型：`torch.float32, torch.float64, torch.float16, torch.uint8, torch.int8, torch.int16, torch.int32, torch.int64` 等。

```python
# 查看 Tensor 的数据类型与形状
x = torch.randn(2, 3)
print('数据类型:', x.dtype)
print('形状:', x.shape)
print('设备:', x.device)
```

## 2. 创建 Tensor 的常用方法

```python
# ones / zeros：全 1 / 全 0
print(torch.ones(2, 3))
print(torch.zeros(2, 3))

# eye：单位矩阵
print(torch.eye(3))

# arange / linspace：等差 / 等间隔序列
print(torch.arange(0, 10))          # [0, 1, ..., 9]
print(torch.linspace(3, 8, 20))     # [3, 8] 之间均匀取 20 个数

# rand / randn：均匀分布 [0,1) / 标准正态分布
print(torch.rand(2, 3))
print(torch.randn(2, 3))

# normal / uniform：指定均值方差的正态 / 指定区间的均匀
print(torch.normal(mean=0, std=1, size=(3,)))
print(torch.empty(3, 3).uniform_(0, 1))

# randperm：随机排列
print(torch.randperm(6))
```

## 3. 定义操作

Tensor 上的各种运算统称为 Function。计算无非三类：基本运算、布尔运算、线性运算。

### 3.1 基本运算（加减乘除、幂、取余、三角函数、取整）

```python
a = torch.tensor([1., 4., 9.])
print('sqrt:', a.sqrt())
print('abs:', torch.tensor([-1., -3.]).abs())
print('exp:', torch.tensor([0., 1.]).exp())
print('pow:', torch.tensor([2., 3.]).pow(2))
print('fmod:', torch.tensor([7., 9.]).fmod(4))   # 取余

print('sin:', torch.tensor([0.]).sin())
print('cos:', torch.tensor([0.]).cos())

# 取整类
print('ceil:', torch.tensor([1.2, 2.8]).ceil())    # 向上取整
print('floor:', torch.tensor([1.2, 2.8]).floor())  # 向下取整
print('round:', torch.tensor([1.2, 2.8]).round())  # 四舍五入
print('trunc:', torch.tensor([1.7, -2.3]).trunc()) # 截断小数部分
```

### 3.2 布尔运算（比较、最大最小、排序）

```python
x = torch.tensor([1, 2, 3, 4])
y = torch.tensor([4, 2, 1, 3])
print('gt:', x.gt(y))   # 大于
print('lt:', x.lt(y))   # 小于
print('eq:', x.eq(y))   # 等于

print('max:', x.max())  # 最大值
print('min:', x.min())  # 最小值

# sort 排序 / topk 前 k 大
print('sort:', x.sort())
print('topk:', x.topk(2))
```

### 3.3 线性运算（矩阵乘法、转置、求模、行列式、逆、奇异值分解）

```python
# 矩阵乘法 mm
A = torch.tensor([[1., 2.], [3., 4.]])
B = torch.tensor([[5., 6.], [7., 8.]])
print('mm:', A.mm(B))          # 矩阵乘

# 转置 t
print('t:', A.t())

# 批量矩阵乘 bmm
batched = torch.randn(10, 2, 3)  # 10 个 2x3 矩阵
batched2 = torch.randn(10, 3, 4)
print('bmm:', batched.bmm(batched2).shape)

# 迹 trace / 对角线 diag
print('trace:', A.trace())
print('diag:', A.diag())

# 点积 dot（要求一维向量）
v1 = torch.tensor([1., 2.])
v2 = torch.tensor([3., 4.])
print('dot:', v1.dot(v2))

# 逆矩阵 inverse / 行列式 det / 奇异值分解 svd
C = torch.tensor([[4., 7.], [2., 6.]])
print('det:', torch.det(C))
print('inverse:', torch.inverse(C))
print('svd:', torch.svd(C))
```

## 小结
- `torch.Tensor` 是 PyTorch 的数据载体（标量、向量、矩阵、高维张量）；
- 创建方式：`ones/zeros/eye/arange/linspace/rand/randn/normal/uniform/randperm` 等；
- 运算：基本运算（`abs/sqrt/exp/pow/fmod/三角函数/取整`）、布尔运算（`gt/lt/eq/max/min/sort/topk`）、线性运算（`mm/bmm/t/dot/trace/diag/inverse/svd`）。
